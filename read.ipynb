```python
import pandas as pd
import numpy as np
```

### Dataset Relationships
These files use two main identifiers: `tconst` for titles and `nconst` for people. `title.basics.tsv` is the central title lookup, while `name.basics.tsv` is the central person lookup. `title.ratings.tsv` and `title.crew.tsv` add title-level details through `tconst`, and `title.principals.tsv` links titles to people through both `tconst` and `nconst`.


### `name.basics.tsv` - People
This table stores one row per person using `nconst` as the person identifier. It connects to `title.principals.tsv` through `nconst`, and its `knownForTitles` values also reference title IDs (`tconst`) from `title.basics.tsv`.


```python
name_basic = pd.read_csv('raw_data/name.basics.tsv', sep='\t', nrows=1000, skiprows=10000, names=['nconst', 'primaryName', 'birthYear', 'deathYear', 'primaryProfession', 'knownForTitles'])
```

```python
name_basic.tail(10)
```

### `title.basics.tsv` - Titles
This is the central title table: each movie, short, episode, or series is identified by `tconst`. Other title-level tables, such as ratings, crew, and principals, link back to this table using the same `tconst` key.


```python
title_basic = pd.read_csv('raw_data/title.basics.tsv', sep='\t', nrows=1000, skiprows=10000, names=['tconst', 'titleType', 'primaryTitle', 'originalTitle', 'isAdult', 'startYear', 'endYear', 'runtimeMinutes', 'genres'])
```

```python
title_basic.tail(10)
```

### `title.crew.tsv` - Directors and Writers
This table lists the main directors and writers for each title. It relates to `title.basics.tsv` by `tconst`; the comma-separated `directors` and `writers` fields contain person IDs (`nconst`) that can be matched to `name.basics.tsv`.


```python
title_crew = pd.read_csv('raw_data/title.crew.tsv', sep='\t', nrows=1000, skiprows=10000, names=['tconst', 'directors', 'writers'])
```

```python
title_crew
```

### `title.principals.tsv` - Main Credits
This table records the principal cast and crew for each title, with one row per credited person and role. It connects titles through `tconst` and people through `nconst`, making it the main bridge between `title.basics.tsv` and `name.basics.tsv`.


```python
title_principals = pd.read_csv('raw_data/title.principals.tsv', sep='\t', nrows=1000, skiprows=10000, names=['tconst', 'ordering', 'nconst', 'category', 'job', 'characters'])
```

```python
title_principals
```

### `title.ratings.tsv` - Ratings
This table stores IMDb rating summary data for each rated title: `averageRating` and `numVotes`. It joins directly to `title.basics.tsv` on `tconst`, adding popularity and rating measures to the title records.


```python
title_ratings = pd.read_csv('raw_data/title.ratings.tsv', sep='\t', nrows=1000, skiprows=10000, names=['tconst', 'averageRating', 'numVotes'])
```
